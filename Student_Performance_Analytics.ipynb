```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Ravi", "Priya", "Kiran", "Sneha"],
    "Department": ["AIML", "CSE", "AIML", "ECE", "CSE"],
    "Python": [85, 72, 90, 65, 78],
    "AI": [88, 68, 95, 70, 82],
    "Maths": [80, 75, 92, 60, 85],
    "Attendance": [92, 85, 98, 72, 90]
}

df = pd.DataFrame(data)

display(df)
```

```python
print(df.columns.tolist())
```

```python
print(df.iloc[0])
```

```python
df["Total_Marks"] = df["Python"] + df["AI"] + df["Maths"]

display(df)
```

```python
df["Average_Marks"] = df["Total_Marks"] / 3

display(df[["Name", "Total_Marks", "Average_Marks"]])
```

```python
def assign_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average_Marks"].apply(assign_grade)

display(df[["Name", "Average_Marks", "Grade"]])
```

```python
def check_result(avg):
    if avg >= 40:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Average_Marks"].apply(check_result)

display(df[["Name", "Average_Marks", "Grade", "Result"]])
```

```python
print("Highest Average:", df["Average_Marks"].max())

print("Lowest Average:", df["Average_Marks"].min())

print("Class Average:", df["Average_Marks"].mean())
```

```python
subjects = ["Python", "AI", "Maths"]

for subject in subjects:
    print(subject, "Average:", df[subject].mean())
```

```python
top_students = df.sort_values(
    by="Average_Marks",
    ascending=False
)

print("Top Performing Students:")
display(
    top_students[
        ["Name", "Average_Marks", "Grade"]
    ].head(3)
)
```

```python
pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()

print("Total Students:", len(df))
print("Passed Students:", pass_count)
print("Failed Students:", fail_count)
```

```python
extra_data = {
    "Student_ID": [106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115,
                   116, 117, 118, 119, 120],
    "Name": ["Arjun", "Meena", "Rahul", "Divya", "Karthik",
             "Pooja", "Sai", "Lakshmi", "Varun", "Nisha",
             "Teja", "Kavya", "Rohit", "Harika", "Manoj"],
    "Department": ["CSE", "AIML", "ECE", "CSE", "AIML",
                   "ECE", "CSE", "AIML", "ECE", "CSE",
                   "AIML", "ECE", "CSE", "AIML", "ECE"],
    "Python": [55, 91, 38, 76, 83, 44, 67, 95, 72, 59,
               88, 63, 31, 79, 86],
    "AI": [60, 94, 42, 81, 79, 50, 70, 89, 75, 62,
           92, 68, 35, 84, 90],
    "Maths": [58, 88, 35, 73, 85, 46, 65, 93, 69, 55,
              84, 71, 40, 77, 82],
    "Attendance": [78, 96, 60, 88, 91, 67, 82, 99, 85, 74,
                   94, 80, 55, 87, 92]
}

extra_df = pd.DataFrame(extra_data)

df = pd.concat([df, extra_df], ignore_index=True)

print("Total Students:", len(df))
display(df)
```

```python
# Calculate total and average marks
subjects = ["Python", "AI", "Maths"]

df["Total_Marks"] = df[subjects].sum(axis=1)
df["Average_Marks"] = df[subjects].mean(axis=1)

# Assign grades
def assign_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average_Marks"].apply(assign_grade)

# Assign Pass or Fail
def check_result(avg):
    if avg >= 40:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Average_Marks"].apply(check_result)

print("Total Students:", len(df))
display(df)
```

```python
# Attendance Analysis

df["Attendance_Status"] = df["Attendance"].apply(
    lambda attendance: "Good Attendance"
    if attendance >= 75
    else "Low Attendance"
)

print("Attendance Analysis:")
display(df[["Name", "Attendance", "Attendance_Status"]])

print("Good Attendance:", (df["Attendance"] >= 75).sum())
print("Low Attendance:", (df["Attendance"] < 75).sum())
```

```python
# Overall Class Performance Analysis

print("Total Students:", len(df))
print("Class Average Marks:", round(df["Average_Marks"].mean(), 2))
print("Highest Average Marks:", round(df["Average_Marks"].max(), 2))
print("Lowest Average Marks:", round(df["Average_Marks"].min(), 2))

print("Total Passed:", (df["Result"] == "Pass").sum())
print("Total Failed:", (df["Result"] == "Fail").sum())

print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")
```

```python
# Department-wise Performance Analysis

department_analysis = df.groupby("Department").agg(
    Total_Students=("Student_ID", "count"),
    Average_Marks=("Average_Marks", "mean"),
    Average_Attendance=("Attendance", "mean")
).round(2)

print("Department-wise Performance:")
display(department_analysis)
```

```python
# Subject-wise Performance Analysis

subjects = ["Python", "AI", "Maths"]

for subject in subjects:
    print(
        subject,
        "Average Marks:",
        round(df[subject].mean(), 2)
    )
```

```python
# Display Top 5 Students

top_5_students = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(5)

print("Top 5 Performing Students:")
display(
    top_5_students[
        ["Student_ID", "Name", "Department",
         "Average_Marks", "Grade"]
    ]
)
```

```python
# Pass/Fail Summary

result_summary = df["Result"].value_counts()

print("Student Result Summary:")
display(result_summary)
```

```python
# Export the final dataset to CSV

df.to_csv("Student_Performance_Analytics.csv", index=False)

print("Dataset exported successfully!")
```
